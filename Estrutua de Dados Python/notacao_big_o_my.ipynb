```python
from math import log
import numpy as np
import matplotlib.pyplot as plt
```

```python
n = np.linspace(1,10,100)
```

```python
labels = ["Constant - k",
          "Logarithmic - log(n)", 
          "Linear - n", 
          "Log Linear - nlog(n)", 
          "Quadratic - n²", 
          "Cubic - n³", 
          "Exponential - 2^n"]

big_o_constant = np.ones(n.shape)
big_o_log = np.log(n)
big_o_linear = n
big_o_log_linear = n*big_o_log
big_o_quadratic = np.power(n,2)
big_o_cubic = np.power(n,3)
big_o_exponential = np.exp2(n)

big_o = [big_o_constant,big_o_log,big_o_linear,big_o_log_linear,big_o_quadratic,big_o_cubic,big_o_exponential]
```

```python
plt.figure(figsize=(10,8))
plt.ylim(0,100)

for i in range(len(big_o)):
    plt.plot(n,big_o[i], label = labels[i])
    
plt.legend()
plt.ylabel("Runtime")
plt.xlabel("n")
```
